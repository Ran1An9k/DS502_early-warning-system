```python
Q1 = df["Semester GPA"].quantile(0.25)
Q3 = df["Semester GPA"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df["Semester GPA"] < lower) | (df["Semester GPA"] > upper)]
```

```python
Q1 = df["Cumulative GPA"].quantile(0.25)
Q3 = df["Cumulative GPA"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df["Cumulative GPA"] < lower) | (df["Cumulative GPA"] > upper)]
```
