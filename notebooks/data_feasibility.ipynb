# Data Feasibility Analysis
Synthetic Academic Dataset for Project 5C

```python
import numpy as np
import pandas as pd

np.random.seed(42)
```

```python
# Basic dataset structure

N_STUDENTS = 500

TERMS = [
    "2023-T1",
    "2023-T2",
    "2024-T1",
    "2024-T2",
    "2025-T1",
    "2025-T2"
]

student_ids = [f"S{i:04d}" for i in range(1, N_STUDENTS + 1)]

df = pd.MultiIndex.from_product(
    [student_ids, TERMS],
    names=["student_id", "term"]
).to_frame(index=False)

print(df.shape)
df.head()
```
