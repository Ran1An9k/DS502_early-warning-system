# Model Comparison

This notebook compares the Logistic Regression baseline model with the Decision Tree candidate model using the same dataset, features, preprocessing steps, train-validation-test split, and evaluation metrics.

## Model Performance Comparison

| Metric | Logistic Regression | Decision Tree |
|---|---:|---:|
| Accuracy | 0.625 | 0.750 |
| Balanced Accuracy | 0.464 | 0.536 |
| Precision | 0.100 | 0.167 |
| Recall | 0.250 | 0.250 |
| F1 Score | 0.143 | 0.200 |
| ROC-AUC | 0.643 | 0.536 |
| PR-AUC | 0.387 | 0.135 |
| Brier Score | 0.130 | 0.250 |

## Comparison Summary

- Decision Tree achieved higher Accuracy, Balanced Accuracy, Precision, and F1 Score.
- Both models achieved the same Recall of 0.25.
- Logistic Regression achieved higher ROC-AUC and PR-AUC.
- Logistic Regression also achieved a lower Brier Score, indicating better probability calibration.
- Both models detected only 1 out of 4 at-risk students in the test set.

## Model Selection

Based on the test results, Logistic Regression was selected as the preferred model.

Although the Decision Tree achieved higher Accuracy and F1 Score, Logistic Regression achieved better ROC-AUC, PR-AUC, and Brier Score.

Because the dataset is imbalanced and the main objective is to identify students at academic risk, PR-AUC and probability quality are important evaluation measures.

Therefore, Logistic Regression was selected as the final model for the current system.
